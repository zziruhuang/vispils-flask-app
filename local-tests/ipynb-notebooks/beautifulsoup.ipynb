```python
from bs4 import BeautifulSoup
```

## Edit Index.html

```python
class HTMLTag():
    def __init__(self, tag, attrs={}, string=''):
        self.bsTag = soup.new_tag(tag, **attrs)
        self.bsTag.string = string

    def appendChildTag(self, childTagList):
        for bT in childTagList:
            self.bsTag.append(bT.bsTag)

        return self
```

```python
cation_families = [
    {   'Name': 'Cyclic amines',
        'Labels': ['[pyr]', '[pip]', '[mor]'],
        'Imgs': ["../static/img/pyr.png", "../static/img/pip.png", "../static/img/mor.png",],
    },
    {   'Name': 'Unsaturated cyclic amines',
        'Labels': ['[im]', '[py]'],
        'Imgs': ["../static/img/pyr.png",  "../static/img/pip.png",],
    },
    {   'Name': 'Acyclic',
        'Labels': ['[s]', '[n]', '[p]'],
        'Imgs': ["../static/img/pyr.png", "../static/img/pip.png", "../static/img/mor.png",]
    }
]
```

```python
with open("templates/index.html") as fp:
    soup = BeautifulSoup(fp, 'html.parser')
print(soup.body)
```

```python
boxOptionListTag = soup.find(class_='box-option-list', id='test-box-bs')
print(boxOptionListTag)
print(boxOptionListTag.contents)
```

```python
for cfam in cation_families:
    childTag = HTMLTag('div', {'class':'option-list'})

    nestedChildTag = HTMLTag('div', {'class':'option'})
    for lb, img in zip(cfam['Labels'], cfam['Imgs'],):
          nestedChildTag.appendChildTag([ 
                  HTMLTag('img', {'src': img}),
                  HTMLTag('p', string=lb)  
              ]
          )
    childTag.appendChildTag([ HTMLTag('h4', string=cfam['Name'],),
                              nestedChildTag]) 
```

```python
boxOptionListTag.append(childTag.bsTag)
```

```python
with open('templates/bs.html', 'w') as file:
    file.write(str(soup.prettify()))
```
