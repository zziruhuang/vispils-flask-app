```python
def pcp_get_smiles(name_list):
    smiles_list=[]
    for mol in name_list:
        result = pcp.get_properties('canonical_smiles', mol, 'name')
        try:
            results = result[0].get(('CanonicalSMILES'))
            print(results)
        except:
            results=[]
            print(mol, 'Error')
        smiles_list.append(results)
        
    return smiles_list
```

```python
def check_name(name):
    df_cation = c_df
    df_anion = a_df
    
    name = name.split(' ')
    cation=[]
    anion =[]
    if len(name) ==2:
        cation = name[0]
        anion = name[1]
    else: 
        cation = name[0]      
        
        name1 = ''.join(name[1:])
        anion = name1
        
    def _look_up_info_file(user_query, df):
        target_lookup = df.loc[(df == user_query).any(axis=1), :]
        input_type = df.loc[:, (df == user_query).any(axis=0)].columns.values
        column_index = df.columns.get_loc(input_type[0])
        target = target_lookup.iloc[0][column_index + 1]
        # row_index = df.loc[(df == user_query).any(axis=1), :].index.tolist()[0]
        return target

    try:
        csmis = _look_up_info_file(cation, df_cation)
        try:
            asmis = _look_up_info_file(anion, df_anion)
        except:
            print('anion failed')
    except:
        print('cation failed')
        
    il = [csmis, asmis]
    ilsmis = '.'.join(il)

    return ilsmis
```
