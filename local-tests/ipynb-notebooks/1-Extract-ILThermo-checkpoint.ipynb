```python
def submit_search(prop='Viscosity'):
    search_button = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((
        By.CSS_SELECTOR, "#sbutton_label"
    )))
    print(search_button.text)
    search_button.click()
    
    ### Click on the ncmp menu ###
    ncmp_menu=driver.find_element(By.CSS_SELECTOR, "#ncmp")
    print(ncmp_menu.text)
    
    if ncmp_menu.get_attribute(name='aria-expanded') == 'false':
        ncmp_menu.click()
        
        ncmp_pure_compound=driver.find_element(By.CSS_SELECTOR,
            "#ncmp_dropdown tr[aria-label='1 - pure compound ']")
        
        if ncmp_pure_compound.get_attribute('aria-selected') == 'false':
            ncmp_pure_compound.click()
    
    ### Click on the prop menu ###
    prop_menu=WebDriverWait(driver,10).until(EC.element_to_be_clickable(
        (By.CSS_SELECTOR, '#prp')
    ))
    
    if prop_menu.get_attribute(name='aria-expanded')=='false':
        prop_menu.click()
        
        prop_tm=WebDriverWait(driver,10).until(EC.element_to_be_clickable(
            (By.CSS_SELECTOR, f"#prp_dropdown tr[aria-label='{prop} ']")
        ))
    
        if prop_tm.get_attribute('aria-selected') == 'false':
            prop_tm.click()
```

```python
# my customized requirement: 
# wait until the text of an element with id='refview' changes
```

```python
class element_change(object):
    def __init__(self, dspanel, ref_content):
        self.dspanel = dspanel
        self.ref_content = ref_content
        
    def __call__(self, driver):
        return self.dspanel.text != self.ref_content
```

```python
class text_present_in_element(object):
    def __init__(self, element, text_list):
        self.element=element
        self.text_list=text_list
        
    def __call__(self, driver):
        for text in self.text_list:  
            if text in self.element.text:
                return True
```

```python
def go_to_stopped_page(stopped_page_num):
    current_page_num=driver.find_element(By.CSS_SELECTOR,
            "[aria-label='Go to page'][class='dgrid-page-link dgrid-page-disabled']")
    current_page_num=int(current_page_num.text)
    print(current_page_num)

    while current_page_num < stopped_page_num:
        page_navigation.click()
        current_page_num=driver.find_element(By.CSS_SELECTOR,
                "[aria-label='Go to page'][class='dgrid-page-link dgrid-page-disabled']")
        current_page_num=int(current_page_num.text)

    print(current_page_num)

    return current_page_num
```

```python
def handle_ds_panel(ds, ref_content, data_headers=['Normal melting temperature, K']):
    ds.click()
    
    dspanel=driver.find_element('id', 'dijit_layout_ContentPane_2')

    WebDriverWait(driver, 120).until(
        element_change(dspanel, ref_content)
    )
    
    dsref=driver.find_element('id','refview')        
    dscomp=driver.find_element('id','dscomp')
    
    f.write(dsref.text)
    f.write('\n')
    f.write(dscomp.text)
    f.write('\n')
    dfcomp_name=dscomp.find_element(By.CSS_SELECTOR,
                'tr:nth-child(2) td:nth-child(2)')
    
    f.write(f"Name: {dfcomp_name.text}")
    f.write('\n') 
    
    dsdata=driver.find_element(By.CSS_SELECTOR, '#dsdata')

    WebDriverWait(driver, 120).until(
        text_present_in_element(dsdata, data_headers))
    f.write(dsdata.text)
    f.write('\n\n')   
    
    ref_data=dsdata.text
    
    return ref_data
    
```
