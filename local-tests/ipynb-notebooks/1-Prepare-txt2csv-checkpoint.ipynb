### Organzing data from Selenium

```python
visData=[]
dvis=[]
with open('ilthermo-vis.txt') as f:
    lines = [line.strip() for line in f]
    
    for line in lines:
        if line != 'Reference':
            dvis.append(line)
        else:
            line = f.readline()
            visData.append(dvis)
            dvis=[]   
f.close()

visData=visData[1:]
```

```python
print(len(visData))
#print(visData[1])
```

```python
titles=[]
citations=[]
published_years=[]
names=[]
methods=[]
phases=[]
headers=[]

ref_ids=[]
allVisData=[]

for i,dvis in enumerate(visData):
    method='Unknown'
    header=''
    dexp=[]
    collect_header=False
    collect_data=False

    try:
        ref_ids.append(i)
        titles.append(dvis[0])
        citations.append(dvis[1])
        published_years.append(int(re.search(r"\((\d+)\)", dvis[1]).group(1)))

        for line in dvis:
            if re.match('^Name: ', line):
                names.append(line.split(": ")[1])
            elif re.match('^Method: ', line):
                method=line.split(': ')[1]
            elif re.match('^Phase\(s\): ', line):
                phases.append(line.split(': ')[1])
            
            if line=='':
                collect_header=True
                collect_data=False
            if re.match('^\d\d', line):
                collect_header=False
                collect_data=True
            
            if collect_header:
                header+= f'{line}\n'
            if collect_data:
                dexp.append(line)
            
        methods.append(method)
        headers.append(header.strip())
        allVisData.append(dexp)
    except:
        print(i)
    
```

```python
propData=[]
dprop=[]
with open('ilthermo-tm.txt') as f:
    lines = [line.strip() for line in f]
    
    for line in lines:
        if line != 'Reference':
            dprop.append(line)
        else:
            line = f.readline()
            propData.append(dprop)
            dprop=[]   
f.close()

propData=propData[1:]
len(propData)
```

```python
titles, citations,published_years=[],[],[]
names=[]
methods, phases, constraints=[],[],[]
headers=[]
ref_ids=[]
allPropData=[]

for i,dcond in enumerate(propData):
    method='Not given'
    header=''
    dexp=[]
    collect_header=False
    collect_data=False

    try:
        ref_ids.append(i)
        titles.append(dcond[0])
        citations.append(dcond[1])
        published_years.append(int(re.search(r"\((\d+)\)", dcond[1]).group(1)))

        for line in dcond:
            if re.match('^Name: ', line):
                names.append(line.split(": ")[1])
            elif re.match('^Method: ', line):
                method=line.split(': ')[1]
            elif re.match('^Phase\(s\): ', line):
                phases.append(line.split(': ')[1])
            elif re.match('^Constraint\(s\): ', line):
                constraints.append(line.split(': ')[1])
            
            if line=='':
                collect_header=True
                collect_data=False
            if re.match('^\d\d', line):
                collect_header=False
                collect_data=True
            
            if collect_header:
                header+= f'{line}\n'
            if collect_data:
                dexp.append(line)
            
        methods.append(method)
        headers.append(header.strip())
        allPropData.append(dexp)
    except:
        print(i)
        
dfInfo = pd.DataFrame({
    "Ref_ID":ref_ids,
    "Title":titles,
    "Citation":citations,
    "Year":published_years,
    "Name":names,
    "Method":methods,
    "Phase":phases,
    "Constrain":constraints,
    "Header":headers
                  })
dfInfo
```

```python
df_allPropData=pd.DataFrame()

propHeaders=['Normal melting temperature, K']
propHeaders_df=[[ph]+['Symbol', 'Uncertainty', 'Name', 'Ref_ID', 'Iso SMILES'] for ph in propHeaders]

for ch, chdf in zip(propHeaders, propHeaders_df):
    print(ch)
    liquidPhases=dfInfo[(dfInfo['Header']==ch) & (dfInfo['Phase']=='Crystal; Liquid')][['Name','Ref_ID','Iso SMILES']]

    propData_passed=[allPropData[i] for i in liquidPhases['Ref_ID']]

    allPropData_passed=[]
    for dprop, dinfo in zip(propData_passed, liquidPhases.values.tolist()):
        dprop_data = [x.split(" ") for x in dprop]
        dprop_data = [x+dinfo for x in dprop_data]
        allPropData_passed.extend(dprop_data)
        
    df_allPropData1=pd.DataFrame(data=allPropData_passed, columns=chdf)
    df_allPropData1=df_allPropData1.drop(columns='Symbol')
    
    for col in [ch]+['Uncertainty']:
        df_allPropData1[col]=[float(x) for x in df_allPropData1[col]]
    
    df_allPropData=pd.concat([df_allPropData, df_allPropData1])
```
