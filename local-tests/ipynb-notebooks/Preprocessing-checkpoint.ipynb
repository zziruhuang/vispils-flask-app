```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df0 = pd.read_csv('Data_extracted_smiles.csv')
df0
```

```python
df = df0[df0['Vicosity Type'].str.match('Viscosity')]
df.columns
```

```python
df_train = df[['Ionic liquid', 'Smiles', 'Temperature (K)', 'Pressure(kPa)', 'Viscosity (mPas)',
       'Log(Viscosity)','Name-cation', 'Smiles-cation','Name-anion','Smiles-anion', ]]
```

```python
df_train
```

```python
duplicateDFRow = df_train[df_train.duplicated()]
duplicateDFRow
```

```python
df_train = df_train.drop_duplicates()
df_train
```

```python
df_train.to_csv('ILthermo_train.csv', index=False)
```

```python
y = df['Log(Viscosity)']
print(len(y))

plt.figure()
n, bins, patches = plt.hist(y, bins=20,
                            #normed = True,
                            align = 'mid',
                            edgecolor = 'black',
                            alpha=0.5,
                           label="Training data")

plt.title("22795 ILs Data", fontsize=16)
plt.xlabel("Log viscosity", fontsize=13)
plt.xticks(fontsize=13)
plt.yticks(fontsize=13)
#plt.legend(loc="best", fontsize=15)
plt.show()
```
